```python
import pickle
with open('polymer_properties_list_qwen3.6-plus(001_100).pkl', 'rb') as f:
    loaded_list_0_100 = pickle.load(f)

with open('polymer_properties_list_qwen3.6-plus(101_300).pkl', 'rb') as f:
    loaded_list_101_300 = pickle.load(f)

with open('polymer_properties_list_qwen3.6-plus(301_580).pkl', 'rb') as f:
    loaded_list_301_580 = pickle.load(f)
```

```python
import io
import re
import pandas as pd

def parse_loaded_list_to_df(loaded_list, start_id):
    all_dataframes = []

    for index, answer_str in enumerate(loaded_list, start=start_id):
        # 改进后的正则：允许最后一行无换行
        table_match = re.findall(r"(?:\|.*?\|\s*(?:\n|$))+", answer_str)
        if table_match:
            table_str = table_match[0]

            # 去掉每行两边的 |，避免 read_csv 产生空列
            table_str = "\n".join([line.strip("|") for line in table_str.strip().splitlines()])

            data_io = io.StringIO(table_str)
            try:
                df = pd.read_csv(data_io, sep="|", skipinitialspace=True, on_bad_lines="warn")
                df.columns = df.columns.str.strip()
                df = df.dropna(how="all")
                df = df.reset_index(drop=True)
                df["ID"] = index
                all_dataframes.append(df)
            except Exception as e:
                print(f"Error parsing entry {index}: {e}")
        else:
            print(f"No table found in entry {index}")

    combined_df = pd.concat(all_dataframes, ignore_index=True)

    return combined_df

df_qwen_0_100 = parse_loaded_list_to_df(
    loaded_list_0_100,
    start_id=1
)

df_qwen_101_300 = parse_loaded_list_to_df(
    loaded_list_101_300,
    start_id=101
)

df_qwen_301_580 = parse_loaded_list_to_df(
    loaded_list_301_580,
    start_id=301
)
```

```python
df_PA_qwen_plus = pd.concat(
    [
        df_qwen_0_100,
        df_qwen_101_300,
        df_qwen_301_580
    ],
    ignore_index=True
)
```

```python
df_PA_qwen_plus = df_PA_qwen_plus[~df_PA_qwen_plus.apply(lambda row: row.astype(str).str.contains('---').any(), axis=1)]
df_cleaned = df_PA_qwen_plus[~df_PA_qwen_plus.apply(lambda row: row.astype(str).str.contains('--').any(), axis=1)]
def remove_temp_unit(x):
    try:
        if 'K' in str(x):
            return float(str(x).replace('K', ''))
        elif '°C' in str(x):
            return float(str(x).replace('°C', ''))
        else:
            return x
    except ValueError:
        return x

df_cleaned['Glass transition temperature'] = df_cleaned['Glass transition temperature'].apply(remove_temp_unit)
df_cleaned=df_cleaned.reset_index(drop=True)
df_cleaned
```

```python
import io
import re
import pandas as pd

def parse_loaded_list_to_df(loaded_list, start_id):
    all_dataframes = []

    for index, answer_str in enumerate(loaded_list, start=start_id):
        # 改进后的正则：允许最后一行无换行
        table_match = re.findall(r"(?:\|.*?\|\s*(?:\n|$))+", answer_str)
        if table_match:
            table_str = table_match[0]

            # 去掉每行两边的 |，避免 read_csv 产生空列
            table_str = "\n".join([line.strip("|") for line in table_str.strip().splitlines()])

            data_io = io.StringIO(table_str)
            try:
                df = pd.read_csv(data_io, sep="|", skipinitialspace=True, on_bad_lines="warn")
                df.columns = df.columns.str.strip()
                df = df.dropna(how="all")
                df = df.reset_index(drop=True)
                df["ID"] = index
                all_dataframes.append(df)
            except Exception as e:
                print(f"Error parsing entry {index}: {e}")
        else:
            print(f"No table found in entry {index}")

    combined_df = pd.concat(all_dataframes, ignore_index=True)

    return combined_df

df_GLM5_0_580 = parse_loaded_list_to_df(
    loaded_list_0_580,
    start_id=1
)
```

```python
import io
import re
import pandas as pd

def parse_loaded_list_to_df(loaded_list, start_id):
    all_dataframes = []

    for index, answer_str in enumerate(loaded_list, start=start_id):
        # 改进后的正则：允许最后一行无换行
        table_match = re.findall(r"(?:\|.*?\|\s*(?:\n|$))+", answer_str)
        if table_match:
            table_str = table_match[0]

            # 去掉每行两边的 |，避免 read_csv 产生空列
            table_str = "\n".join([line.strip("|") for line in table_str.strip().splitlines()])

            data_io = io.StringIO(table_str)
            try:
                df = pd.read_csv(data_io, sep="|", skipinitialspace=True, on_bad_lines="warn")
                df.columns = df.columns.str.strip()
                df = df.dropna(how="all")
                df = df.reset_index(drop=True)
                df["ID"] = index
                all_dataframes.append(df)
            except Exception as e:
                print(f"Error parsing entry {index}: {e}")
        else:
            print(f"No table found in entry {index}")

    combined_df = pd.concat(all_dataframes, ignore_index=True)

    return combined_df

df_gpt54mini_0_580 = parse_loaded_list_to_df(
    polymer_properties_list,
    start_id=1
)
```

```python
df_gpt54mini_0_580 = df_gpt54mini_0_580[~df_gpt54mini_0_580.apply(lambda row: row.astype(str).str.contains('---').any(), axis=1)]
df_cleaned = df_gpt54mini_0_580[~df_gpt54mini_0_580.apply(lambda row: row.astype(str).str.contains('--').any(), axis=1)]
def remove_temp_unit(x):
    try:
        if 'K' in str(x):
            return float(str(x).replace('K', ''))
        elif '°C' in str(x):
            return float(str(x).replace('°C', ''))
        else:
            return x
    except ValueError:
        return x

df_cleaned['Glass transition temperature'] = df_cleaned['Glass transition temperature'].apply(remove_temp_unit)
df_cleaned=df_cleaned.reset_index(drop=True)
df_cleaned
```
